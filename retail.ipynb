```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('menu.csv')
df.head(10)
```

```python
df.info()
```

```python
df.isnull().sum()

```

```python
#there is no missing values
#data types are already correct
#no further cleanig is required
# Check duplicates
print("Duplicate Rows:", df.duplicated().sum())

# Remove duplicates if present
df = df.drop_duplicates()

# Save cleaned dataset
df.to_csv("menu_cleaned.csv", index=False)

print("Cleaned dataset saved successfully")
```

```python
df.describe()
```

```python

num_cols = df.select_dtypes(include='number').columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns
cat_cols = df.select_dtypes(include='object').columns

for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#first categoory distrubution
plt.figure(figsize=(10,6))

sns.countplot(
    y=df['Category'].value_counts().index
)
plt.title("Number of items in Each category for MCdonalds")

plt.show()
```

```python
#second step
#calories distrubution
plt.figure(figsize=(10,6))
 
sns.histplot(df['Calories'], bins=30, kde=True)
plt.title("Calories distribution for MCdonalds")
```

```python
#top calories items
top_calories = df.nlargest(10, 'Calories')

top_calories[['Item','Calories']]
```

```python
#visualization of top 10 high calorie items
plt.figure(figsize=(10,6))
 
sns.barplot(
    x='Calories',
    y='Item',
    data=top_calories
 
 )
plt.title("Top 10 High-Calorie Items in McDonald's Menu")
```

```python
#average Calorries by each category
avg_calories=df.groupby('Category')['Calories'].mean()
avg_calories.sort_values(ascending=False)
plt.figure(figsize=(10,6))
#chart for average calories by category
sns.barplot(
    x=avg_calories.values,
    y=avg_calories.index
)
plt.title("Average Calories by Category")
plt.show()
```

```python
#fat analysis
total_fat=df.groupby('Category')['Total Fat'].mean()
total_fat.sort_values(ascending=False)
plt.figure(figsize=(10,6))
sns.barplot(
    x=total_fat.values,
    y=total_fat.index   
)
plt.title("Average Total Fat by Category")
plt.show()
```

```python
#protein analysis
gain_protein=df.groupby('Category')['Protein'].mean()
gain_protein.sort_values(ascending=False)   

plt.figure(figsize=(10,6))
sns.barplot(    
    x=gain_protein.values,
    y=gain_protein.index
)
plt.title("  top  10 Protein  items by  each Category")
plt.show()  

```

```python
high_protein_items=df.nlargest(10,'Protein')
high_protein_items[['Item','Protein']]  
plt.figure(figsize=(10,6))
sns.barplot(
    x='Protein',
    y='Item',
    data=high_protein_items
)
plt.title("High Protein Items in McDonald's Menu")
plt.show()
```

```python
# correlation analysis
#nutritional correlation matrix
numeric_df = df.select_dtypes(include=['number'])
correlation_matrix = numeric_df.corr()
plt.figure(figsize=(12,8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Nutritional Correlation Matrix")
plt.show()
```

```python
vitamins=df.groupby("Category")[['Vitamin A (% Daily Value)', 'Vitamin C (% Daily Value)','Calcium (% Daily Value)', 'Iron (% Daily Value)']].mean()
#vitamin content heatmap
plt.figure(figsize=(10,6))
sns.heatmap(
    vitamins,
    annot=True,
    cmap='YlGnBu',
    fmt='.2f'
)
plt.title("Average Vitamin Content by Category")
plt.show()
```

```python
risk_data=df.groupby('Category')[['Sodium','Sugars','Total Fat']].mean()
#correlation for the sodium,sugars,fat risk factors
plt.figure(figsize=(10,6))  
sns.heatmap(
    risk_data,
    annot=True,
    cmap='Reds',
    fmt='.2f'
)
plt.title("Average Risk Factors by Category")
plt.show()
```

```python
#calories vs protein  correaltion 
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='Protein',
    y='Calories'
)

plt.title('Calories vs Protein')
plt.xlabel('Protein (g)')
plt.ylabel('Calories')

plt.show()
```

```python
#correlation for calories vs Fat
plt.figure(figsize=(10,6))
sns.scatterplot(
    x='Calories',
    y='Total Fat',
    data=df
)
plt.title("Calories vs Total Fat")
plt.show()
```

```python

#calories vs sugars
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='Sugars',
    y='Calories'
)

plt.title('Calories vs Sugars')
plt.xlabel('Sugar (g)')
plt.ylabel('Calories')

plt.show()
```

```python
#healthy food score
df['Healthy Score'] = (
    df['Protein']
    - df['Total Fat']
    - df['Sugars']
)

df[['Item','Healthy Score']].sort_values(by='Healthy Score', ascending=True).head(10)
```

```python
#Top healthy food items
healthy_food=df.nlargest(10,'Healthy Score')
healthy_food[['Item','Healthy Score']]
 
 
```

```python
#the chart for the healthy items
plt.figure(figsize=(10,6))
sns.barplot(
    x='Healthy Score',
    y='Item',
    data=healthy_food
)
plt.title("Top 10 Healthy Items in McDonald's Menu")
```

```python
nutrition_summary = df.groupby('Category')[
    ['Calories','Total Fat','Sugars','Protein']
].mean()

nutrition_summary.sort_values(by='Calories', ascending=False)
```
